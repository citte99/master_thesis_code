```python
import os
import time
import torch
from torch.utils.data import DataLoader, TensorDataset
from tqdm import tqdm
from deep_learning.NN_datasets import NoNoiseDataset
from deep_learning.NN_datasets.dataloaders import custom_dataloader

# -------------------- Configuration --------------------
device      = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
batch_size  = 1024
num_images  = 100_000   # total samples to test (~100 batches)
cache_path  = 'raw_cache.pt'

# -------------------- On-the-fly loader --------------------
base_ds     = NoNoiseDataset(
    "min_mass_10e10",
    grid_pixel_side=80,
    grid_width_arcsec=8.0,
    broadcasting=True,
    samples_used=num_images,
    upscaling=5
)
loader_otf  = custom_dataloader(base_ds, batch_size=batch_size)

# -------------------- Cache raw images --------------------
if not os.path.exists(cache_path):
    imgs = []
    lbls = []
    # Estimate number of batches
    num_batches = (num_images + batch_size - 1) // batch_size
    for src, tgt in tqdm(loader_otf, desc="Caching raw images", total=num_batches):
        imgs.append(src)    # raw, no noise/normalize
        lbls.append(tgt)
        if len(imgs) >= num_batches:
            break
    images = torch.cat(imgs, dim=0)[:num_images]
    labels = torch.cat(lbls, dim=0)[:num_images]
    torch.save((images, labels), cache_path)

# -------------------- Cached loader --------------------
images, labels = torch.load(cache_path, map_location='cpu')
cached_ds       = TensorDataset(images, labels)
loader_cached   = DataLoader(
    cached_ds,
    batch_size=batch_size,
    shuffle=True,
    num_workers=0,
    pin_memory=True
)

# -------------------- Benchmark loader speeds --------------------
def benchmark_loader(loader, desc):
    it = iter(loader)
    # Warm-up
    for _ in range(5):
        x, _ = next(it)
        _ = x.to(device, non_blocking=True)
    torch.cuda.synchronize()

    # Timed
    it = iter(loader)
    n = (num_images + batch_size - 1) // batch_size
    start = time.time()
    for _ in tqdm(range(n), desc=desc):
        x, _ = next(it)
        _ = x.to(device, non_blocking=True)
    torch.cuda.synchronize()
    return (time.time() - start) / n

t_otf   = benchmark_loader(loader_otf,   "On-the-fly load")
t_cache = benchmark_loader(loader_cached, "Cached load")

print(f"Avg load time (on-the-fly): {t_otf:.4f} s/batch")
print(f"Avg load time (cached):    {t_cache:.4f} s/batch")
```
