```python
import matplotlib.pyplot as plt
import torch
import time
import numpy as np

# Load your modules and create iterator
module('load', 'pytorch/gpu-cuda-12.1/2.2.0')
from deep_learning import NoNoiseDataset, custom_dataloader

dataset = NoNoiseDataset(
    catalog_name = 'min_mass_10e11_test',
    grid_pixel_side = 80, 
    grid_width_arcsec = 8.,
    broadcasting= True ,
    upscaling = 5
)
dataloader = custom_dataloader(dataset, 1024)
iterator = iter(dataloader)

# Performance comparison
n_iterations = 100

# Timing arrays
optical_times = []
interferometric_times = []

print(f"Running performance comparison for {n_iterations} iterations...")

for i in range(n_iterations):
    try:
        image = next(iterator)[0]
    except StopIteration:
        # Reset iterator if we run out of data
        iterator = iter(dataloader)
        image = next(iterator)[0]
    
    # Time optical noiser
    torch.cuda.synchronize()
    start = time.time()
    image_optical = noiser(image)
    torch.cuda.synchronize()
    optical_time = time.time() - start
    optical_times.append(optical_time)
    
    # Time interferometric noiser
    torch.cuda.synchronize()
    start = time.time()
    image_interf = noiser_fft(image)
    torch.cuda.synchronize()
    interf_time = time.time() - start
    interferometric_times.append(interf_time)
    
    # Show a few examples
    if i < 3:
        fig, axes = plt.subplots(1, 3, figsize=(15, 5))
        
        im0 = axes[0].imshow(image[0][0].cpu().numpy())
        axes[0].set_title('Original Image')
        plt.colorbar(im0, ax=axes[0])
        
        im1 = axes[1].imshow(image_optical[0][0].cpu().numpy())
        axes[1].set_title('Optical PSF + Poisson')
        plt.colorbar(im1, ax=axes[1])
        
        im2 = axes[2].imshow(image_interf[0][0].cpu().numpy())
        axes[2].set_title('Interferometric PSF + Fourier noise')
        plt.colorbar(im2, ax=axes[2])
        
        plt.tight_layout()
        plt.show()

# Performance summary
print("\nPerformance Summary:")
print(f"Batch size: {image.shape[0]}")
print(f"Image size: {image.shape[-2:]}") 
print(f"Optical Noiser: {np.mean(optical_times)*1000:.2f} ± {np.std(optical_times)*1000:.2f} ms")
print(f"Interferometric Noiser: {np.mean(interferometric_times)*1000:.2f} ± {np.std(interferometric_times)*1000:.2f} ms")
print(f"Slowdown factor: {np.mean(interferometric_times)/np.mean(optical_times):.2f}x")

# Timing plot
plt.figure(figsize=(10, 6))
plt.plot(optical_times, label='Optical Noiser', alpha=0.7)
plt.plot(interferometric_times, label='Interferometric Noiser', alpha=0.7)
plt.xlabel('Iteration')
plt.ylabel('Time (seconds)')
plt.title('Noiser Performance Comparison')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# Histogram
plt.figure(figsize=(10, 6))
plt.hist(np.array(optical_times)*1000, bins=30, alpha=0.5, label='Optical')
plt.hist(np.array(interferometric_times)*1000, bins=30, alpha=0.5, label='Interferometric')
plt.xlabel('Time per batch (ms)')
plt.ylabel('Count')
plt.title('Timing Distribution')
plt.legend()
plt.show()
```

```python
import pandas as pd

dict_conf = full_sys_conf.get_pth()

def tensor_dict_to_dataframes(nested_dict):
    dfs = {}
    for key, subdict in nested_dict.items():
        # top‐level component, e.g. flat['precomputed']
        if isinstance(subdict, dict) and 'params' in subdict and 'param_map' in subdict:
            arr = subdict['params'].cpu().numpy()
            dfs[key] = pd.DataFrame(arr, columns=subdict['param_map'])
        # second level, e.g. flat['mass_components']['PEMD']
        elif isinstance(subdict, dict):
            for subkey, inner in subdict.items():
                if isinstance(inner, dict) and 'params' in inner and 'param_map' in inner:
                    arr = inner['params'].cpu().numpy()
                    name = f"{key}_{subkey}"
                    dfs[name] = pd.DataFrame(arr, columns=inner['param_map'])
    return dfs


dfs = tensor_dict_to_dataframes(dict_conf)
```

```python
import matplotlib.pyplot as plt
import numpy as np

for name, df in dfs.items():
    for col in df.columns:
        values = df[col].values
        finite_vals = values[np.isfinite(values)]
        if len(finite_vals) == 0:
            print(f"Skipping {name}.{col} — all values are non-finite.")
            continue
        plt.figure()
        plt.hist(finite_vals, bins=50)
        plt.title(f"{name} — {col}")
        plt.xlabel(col)
        plt.ylabel("Frequency")
        plt.tight_layout()
        plt.show()
```
